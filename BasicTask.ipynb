**2. Data Handling with Pandas**

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("miadul/introvert-extrovert-and-ambivert-classification")

print("Path to dataset files:", path)
```

```python
a=os.listdir(path)
print(a)
```

```python
data=pd.read_csv(path+'/personality_synthetic_dataset.csv')
data.head(10)
```

```python
#store data in DataFrame
data=pd.DataFrame(data)
```

```python
data.info()
```

```python
data.describe()
```

```python
data.isnull().sum()
```

```python
data.info()
```

```python
encoder=LabelEncoder()
data["personality_type"]=encoder.fit_transform(data["personality_type"])
```

```python
data.head(5)   #extrovert-1, introvert-2, Ambrivort-0
```

```python
X=data.drop('personality_type',axis=1)
Y=data["personality_type"]
```

```python
print("Shpae of X::",X.shape)
print("Shape of Y::",Y.shape)
```

**Data Visulization**

```python
#matplotlib.pyplot

#line plot
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y, color='blue', linestyle='--', marker='o', label='Line')
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()
```

```python
#Bar plot
courses = ["OOP","DSA","AOA","ML"]
enroll_count = [23,10,30,10]
color=["red","green","blue","yellow"]

plt.bar(courses,enroll_count, color=color)  #by default vertical
plt.title("Bar Chart")
plt.xlabel("Courses")
plt.ylabel("enrollment_count")
plt.show()
```

```python
#horizental bar chart
plt.barh(courses,enroll_count, color=color)
plt.title("Horizontal Bar Chart")
plt.xlabel("courses")
plt.ylabel("enrollment_count")
plt.show()
```

```python
#histogram
plt.hist(X.iloc[:,0],bins=30,color="skyblue",edgecolor="black")
plt.title("Histogram")
plt.xlabel("Time_spent_Alone")
plt.ylabel("Frequency")
plt.show()
```

```python
#scatter plot
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y, color='red', marker='x')
plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()
```

```python
#pie plot


plt.pie(enroll_count,labels=courses, colors=color, autopct='%1.1f%%', startangle=90)
plt.title("Pie Chart")
plt.axis('equal') #to make sure equal ratio , so circle properly drawn
plt.show()
```

```python
#Box plot



plt.boxplot(X)
plt.title("Box Plot")
plt.xlabel("Groups")
plt.ylabel("Values")
plt.show()

```

```python
a= np.random.rand(5, 5)

plt.imshow(a, cmap='viridis', interpolation='nearest')
plt.colorbar()
plt.title("Heatmap using imshow")
plt.show()
```

```python
#subplot

fig, axs = plt.subplots(1, 2, figsize=(10, 5))

axs[0].scatter(X.iloc[:,1], Y, color='blue')
axs[0].set_title('Time_spent_Alone')
axs[0].set_xlabel('Alone_Time')
axs[0].set_ylabel('Personality')

axs[1].scatter(X.iloc[:,2], Y, color='green')
axs[1].set_title('Talkativeness')
axs[1].set_xlabel('Talkativeness_rate')
axs[1].set_ylabel('Personality')
plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(data=data, x="listening_skill", y="group_comfort", hue="personality_type")
plt.title("Scatter Plot")
plt.show()
```

```python
#line plot
sns.lineplot(data=data,x="social_energy", y="sports_interest", hue="personality_type" )
plt.title("Line Plot")
plt.show()
```

```python
#bar plot

data['social_usage_rounded'] = data['online_social_usage'].round()
sns.barplot(data=data, x='social_usage_rounded', y='personality_type', ci='sd')
plt.title("Personality Type vs Online Social Usage")
plt.show()
```

```python
#box plot
data['work_style_collaborative'] = data['work_style_collaborative'].round()
sns.boxplot(data=data, x="work_style_collaborative", y="decision_speed", hue="personality_type")
plt.title("Box Plot")
plt.show()
```

```python
#violin plot
sns.violinplot(data=data, x="personality_type", y="stress_handling", hue="personality_type", palette="Set2")
plt.title("Violin Plot of Stress Handling by Personality Type")
plt.show()
```

```python
#heatmap
sns.heatmap(data.corr(), annot=False, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
